class HybridFusionModel(nn.Module):
    def __init__(self, gesture_dim=33*3, context_dim=10, hidden_dim=128, num_classes=2):
        super().__init__()
        self.lstm = nn.LSTM(gesture_dim, hidden_dim, num_layers=2, batch_first=True)  # Process gesture sequences
        self.transformer = nn.TransformerEncoder(nn.TransformerEncoderLayer(d_model=context_dim, nhead=2), num_layers=2)  # Weight contexts
        self.fc = nn.Linear(hidden_dim + context_dim, num_classes)  # Fuse and classify

        trans_out = self.transformer(contexts.unsqueeze(1))  # Add seq dim if needed, output [batch, 1, context_dim]

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
import numpy as np

# --- 1. Generate SYNTHETIC (Fake) Data ---
# Imagine we have 1000 samples
# Gestures: sequence of 20 frames, 10 features (e.g., x,y coordinates of body parts)
# Contexts: 3 features (e.g., Is_Raining? Is_Night? Traffic_Density?)
num_samples = 1000
seq_length = 20
gesture_features = 10
context_features = 3

# Random data generation
X_gestures = np.random.randn(num_samples, seq_length, gesture_features).astype(np.float32)
X_contexts = np.random.randn(num_samples, context_features).astype(np.float32)
# Labels: 0 (Won't Cross) or 1 (Will Cross)
y_labels = np.random.randint(0, 2, size=(num_samples)).astype(np.int64)

# --- 2. Define the Dataset Class ---
class PedestrianDataset(Dataset):
    def __init__(self, gestures, contexts, labels):
        self.gestures = torch.tensor(gestures)
        self.contexts = torch.tensor(contexts)
        self.labels = torch.tensor(labels)

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, idx):
        return self.gestures[idx], self.contexts[idx], self.labels[idx]

# Create DataLoader
dataset = PedestrianDataset(X_gestures, X_contexts, y_labels)
train_loader = DataLoader(dataset, batch_size=32, shuffle=True)

# --- 3. Define the Novel Model (Fusion) ---
class FusionModel(nn.Module):
    def __init__(self):
        super(FusionModel, self).__init__()
        # LSTM for Gestures (Temporal)
        self.lstm = nn.LSTM(input_size=10, hidden_size=64, batch_first=True)

        # Simple Linear Layer for Context (Environmental)
        self.context_layer = nn.Linear(3, 16)

        # Fusion Layer (Concatenate LSTM output + Context output)
        self.fusion_layer = nn.Linear(64 + 16, 32)
        self.output_layer = nn.Linear(32, 2) # Binary classification

    def forward(self, gestures, contexts):
        # Process gestures
        _, (hidden, _) = self.lstm(gestures)
        lstm_out = hidden[-1] # Take the last hidden state [Batch, 64]

        # Process context
        context_out = torch.relu(self.context_layer(contexts)) # [Batch, 16]

        # Fuse
        combined = torch.cat((lstm_out, context_out), dim=1) # [Batch, 80]
        x = torch.relu(self.fusion_layer(combined))
        return self.output_layer(x)

# --- 4. Train ---
model = FusionModel()
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

print("Starting Training on Synthetic Data...")
for epoch in range(5):
    total_loss = 0
    for gestures, contexts, labels in train_loader:
        optimizer.zero_grad()
        outputs = model(gestures, contexts)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
        total_loss += loss.item()
    print(f"Epoch {epoch+1}, Loss: {total_loss/len(train_loader):.4f}")

print("Success! The model architecture works.")
```

```python
# --- ONLY RUN THIS AFTER YOU HAVE UPLOADED AND EXTRACTED NUSCENES ---

import numpy as np
from nuscenes.nuscenes import NuScenes
from nuscenes.utils.splits import create_splits_scenes

class RealNuScenesDataset(Dataset):
    def __init__(self, nusc, seq_len=10):
        self.nusc = nusc
        self.seq_len = seq_len
        self.samples = []

        print("Indexing dataset... this may take a minute.")
        self._build_index()

    def _build_index(self):
        # 1. Iterate through every sample (snapshot in time)
        for sample in self.nusc.sample:
            scene_token = sample['scene_token']
            scene = self.nusc.get('scene', scene_token)
            description = scene['description'].lower()

            # --- EXTRACT CONTEXT (The Novelty) ---
            # Simple keyword matching for environment
            is_night = 1 if 'night' in description else 0
            is_rain = 1 if 'rain' in description else 0
            # Context Vector: [Night, Rain]
            context = [is_night, is_rain]

            # 2. Find Pedestrians in this sample
            for ann_token in sample['anns']:
                ann = self.nusc.get('sample_annotation', ann_token)

                # Filter: Only look at humans
                if ann['category_name'].startswith('human.pedestrian'):

                    # 3. Get Trajectory (History)
                    # We need to look BACKWARDS in time to get a sequence
                    history = self._get_history(ann, self.seq_len)

                    # Only keep if we have enough history (full sequence)
                    if len(history) == self.seq_len:
                        # Create a data entry
                        # Label: For now, let's try to predict if they are moving (speed > 0.5)
                        # In real research, we would look at future attributes for "crossing"
                        # For this prototype: 1 if moving, 0 if standing still
                        speed = np.linalg.norm(self.nusc.get('sample_annotation', ann['next'])['translation'][:2]) if ann['next'] else 0
                        label = 1 if speed > 0.5 else 0

                        self.samples.append({
                            'gesture': np.array(history), # Shape: [seq_len, 3] (x,y,z coords)
                            'context': np.array(context), # Shape: [2]
                            'label': label
                        })

    def _get_history(self, current_ann, depth):
        """ helper function to trace back previous frames """
        traj = []
        curr = current_ann
        for _ in range(depth):
            # Get x,y,z coordinates
            traj.append(curr['translation'])

            # Go to previous frame
            if curr['prev'] == '':
                break
            curr = self.nusc.get('sample_annotation', curr['prev'])

        # We want order: Oldest -> Newest
        return traj[::-1]

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        item = self.samples[idx]
        # Return tensors ready for the model
        return (
            torch.tensor(item['gesture'], dtype=torch.float32),
            torch.tensor(item['context'], dtype=torch.float32),
            torch.tensor(item['label'], dtype=torch.long)
        )

# --- HOW TO USE THIS ---
# 1. Load NuScenes (Make sure you did the extraction step first!)
# nusc = NuScenes(version='v1.0-mini', dataroot='/content/nuscenes_mini', verbose=True)

# 2. Create Dataset
# real_dataset = RealNuScenesDataset(nusc, seq_len=5)
# print(f"Found {len(real_dataset)} usable pedestrian sequences.")

# 3. Create Loader
# train_loader = DataLoader(real_dataset, batch_size=32, shuffle=True)

# 4. Update Model Input Sizes
# The LSTM input size is now 3 (x, y, z coordinates), not 10.
# The Context input size is now 2 (Rain, Night), not 3.
# model = FusionModel(input_size=3, context_size=2) ...
```
